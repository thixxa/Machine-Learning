```python
import numpy as np
data = np.array([[28, 70000],
                 [49, 90000],
                 [24, 25000],
                 [41, 31000]])
```

```python
# Normalization
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
scaled_data
```

```python
# Standardization
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
scaled_data
```
