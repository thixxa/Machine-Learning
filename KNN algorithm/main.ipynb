```python
# feature tika gannawa x widihata
# iloc kiyanne index location
x = data.iloc[:,1:5] 
x.head()
```

```python
y = data.iloc[:,-1] # -1 kiyala ganne last column eka
y.head()
```

```python
# feature scaling
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x = scaler.fit_transform(x)
x[0:5] # x wala mul value 5 balanawa
```

```python
# Train test split
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
# K-NN algorithm
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=1) # k = 1 kiyala dila (mulin 1 kiyala nikan denawa)
model.fit(x_train, y_train)
```

```python
pred = model.predict(x_test) # testing eke data tikakata predict karnawa
pred[0:5] # predict karapu eke mul 5
```

```python
y_test[0:5]
```

```python
# To find accuracy
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, pred)
accuracy
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, pred) # wikarnaye thiyena agayan thama true tika
cm
```

```python
result = pd.DataFrame(data=[y_test.values,pred], index=['y_test','pred'])
result.transpose()
```

```python
# To identify correct K value
```

```python
correct_sum = [] 
for i in range(1,20):
    model = KNeighborsClassifier(n_neighbors=i)
    model.fit(x_train,y_train)
    pred = model.predict(x_test)
    correct = np.sum(pred == y_test)
    correct_sum.append(correct)
```

```python
correct_sum
```

```python
result = pd.DataFrame(data=correct_sum)
result.index = result.index+1
result.T
```

```python
# K = 3 ganna
```

```python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(x_train,y_train)
pred = model.predict(x_test)
```

```python
accuracy_score(y_test,pred)
```

```python
confusion_matrix(y_test,pred)
```
